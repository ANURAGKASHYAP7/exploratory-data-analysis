```python
import pandas as pd
import numpy as np
df = pd.read_csv("House Price India.csv")
df
```

```python
'''Q1. What is the average, median, and standard deviation of house prices?
'''
df["Price"].agg(["mean","median","std"])
```

```python
'''
Q2. Which number of bedrooms is most common?
'''
df["number of bedrooms"].mode()
```

```python
df["number of bedrooms"].value_counts()
```

```python
'''. Q3 Check if the price data is skewed
'''
df["Price"].skew()
```

 Q3 The price data is positively skewed (right-skewed), indicating that some houses have very high prices compared to the majority of houses.

```python
'''Q4. What is the average price for each number of bedrooms?
'''
df.head()
df.groupby("number of bedrooms")["Price"].mean()
```

```python
'''Q5.  What is the relationship between living area and price? (Use : Area without Basement).
'''
df["Area of the house(excluding basement)"].corr(df["Price"])
#The correlation between living area (excluding basement) and price is 0.615, indicating a positive relationship. As the living area increases, the house price generally tends to increase.
```

```python
'''Q6. Identify any anomalies where houses have high prices but low areas.
'''
df["Area of the house(excluding basement)"].median()
df["Price"].median()
df[(df["Area of the house(excluding basement)"] < df["Area of the house(excluding basement)"].median()) &
   (df["Price"] > df["Price"].median())]
```

```python
'''Q7. Compare average price based on number of floors and houses with or without waterfront
'''
df.groupby("number of floors")["Price"].mean()
```

```python
df.groupby("waterfront present")["Price"].mean()
```

```python
'''Q8. Identify the minimum and maximum house price. What does this indicate?
'''
df["Price"].min()
df["Price"].max()
#The minimum and maximum house prices show that house prices vary over a wide range, indicating considerable variation in house prices.
```

```python
'''9. Which location (zipcode/area) has the highest average price?


'''
df.groupby("Postal Code")["Price"].mean()
```
